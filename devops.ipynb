1. What is DevOps? Explain its goals and benefits.

   - Devops is set of practices and collection of tools that help development and opeartions team work together throughout the software development life cycle to deliver software faster,more reliable and with higher quality.

   Goals of DevOps-
      a. Faster Software Delivery.
      b. Better Collaborations - Developers and operations teams work together instead of separately.
      c. Automation - Reduce manual work by automating by testing, building, deployment, monitoring.
      d. Continuous Integration and Continuous Deployment.
      e. Faster Recovery that minimize downtime.

  Benefits of DevOps-
      a. Faster Deployment
      b. Better Quality - Because of automated testing and code reviews very few bugs reach production.
      c. Improved Collaboration
      d. Lower Failure Rate - Smaller, frequent deployments are easier to verify and fix.
      e. Faster Bug Fixes - Bug fix and deploy immediately.
      f. Cost Savings - it reduces Human effort and Operational costs.


2. Explain the history and evolution of DevOps.

   - DevOps evolved to solve the communication gap between Development and Operations teams and to enable faster, more reliable software delivery.

  Before 2000: Developers and operations teams worked separately, leading to slow releases, manual deployments, and frequent errors.

  2001 - Agile: The Agile methodology introduced faster development and continuous feedback but did not fully address deployment challenges.

  2009 - Birth of DevOps: The term DevOps became popular after Patrick Debois organized the first DevOpsDays conference, promoting collaboration between development and operations teams.
  
  Modern DevOps: Today, DevOps uses automation, CI/CD, cloud computing, containers like Docker, and orchestration tools like Kubernetes to deliver software quickly, reliably, and securely.

3. What is SDLC? Explain traditional development models.
    - SDLC (Software Development Life Cycle) is a structured process used to develop high-quality software. It includes stages such as Planning, Analysis, Design, Development, Testing, Deployment, and Maintenance.

  Traditional Development Models are -
  a. Waterfall Model :
      This is linear and step-by-step approach.
      Each phase must be completed before moving to the next.
      Best for projects with fixed requirements.
      
  b. V-Model :
      An extension of the Waterfall model.
      Testing is planned alongside each development phase.
      Ensures better quality through early testing.
      
  c. Spiral Model :
      Combines development with risk analysis.
      Software is developed in repeated cycles.
      Suitable for large and high-risk projects.

4. Explain Agile methodologies: Scrum and Kanban.
    - Agile is a software development methods that focuses on flexibility, collaboration, continuous feedback, and quick delivery.

  a. Scrum:
      -Work is divided into short Sprints (usually 2–4 weeks).
      -Team members have defined roles such as Scrum Master, Product Owner, and Development Team.
      -Daily Scrum meetings are held to track progress.
      -Best for projects with changing requirements.

  b. Kanban:
      -Work is managed using a Kanban Board with columns like To Do, In Progress, and Done.
      -Tasks move across the board as work progresses.
      -Focuses on continuous delivery.

5. Explain the role of an Operating System and the difference between processes and threads, process lifecycle.

    - An Operating System is system software that acts as an interface between the user and computer hardware. Its main functions are:

     a. Manages CPU, memory, files, and devices.
     b. Runs and controls programs.
     c. Provides security and user interface.

      Difference Between Process and Thread -
      
      Process	                                Thread
      a.An independent program in execution.	a. A smaller unit of a process.
      b. Has its own memory space.	          b. Shares memory with other threads of the same process.
      c. More resource-intensive.	            c.Lightweight and faster.


      Process Lifecycle -

      A process passes through the following states:

      a. New : Process is created.
      b. Ready : Waiting for CPU.
      c. Running : Executing on the CPU.
      d. Waiting/Blocked : Waiting for input/output or an event.
      e. Terminated : Process finishes execution.

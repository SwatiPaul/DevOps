1. What is process scheduling? Explain multitasking, time-slicing,
and priority scheduling.
   - Process Scheduling is the function of the Operating System that decides which process gets the CPU and for how long. It helps improve CPU utilization and system performance.

      a. Multitasking:
      Allows multiple programs to run at the same time.
      The CPU switches quickly between processes, giving the impression that they run simultaneously.

      b. Time-Slicing:
      Each process is given a fixed amount of CPU time called a time slice or time quantum.
      After its time expires, the CPU switches to the next process.
      
      c. Priority Scheduling:
      Each process is assigned a priority.
      The process with the highest priority is executed first.
      Processes with lower priority may have to wait longer.

2. What is virtual memory, paging, and segmentation?

  - Virtual Memory is a memory management technique that uses part of the hard disk as temporary memory when RAM is full.The OS Stores extra memory on the hard disk (called swap space or page file).

  Paging is a memory management technique where Memory is divided into fixed-size blocks called pages.
  Physical memory is divided into frames and Pages are mapped to frames, reducing external fragmentation.
  Instead of needing one large empty space, the OS can place pages into any free frames.

  Segmentation divides a program into variable-sized segments based on logical units such as Code, Data and Stack.
  Each segment is stored separately in memory, making memory management more flexible.

3. What is memory fragmentation and swap memory?

      - Memory Fragmentation occurs when free memory is broken into small pieces, making it difficult to allocate memory to a program.
      eg - if want to place a large book it needs a continuous space. Even though enough total space exists, it cannot fit because the free space is fragmented.

      There are 2 types of Fragmentation -
      Internal Fragmentation:
      Wasted space inside an allocated memory block.

      External Fragmentation:
      Free memory exists but is scattered in small blocks.
      A large program cannot be allocated because there is no single continuous block of memory.

      Swap Memory - Swap Memory is a portion of the hard disk (or SSD) used as temporary memory when RAM becomes full.

      The OS moves inactive data from RAM to the swap area on the disk.
      When needed again, the data is moved back to RAM.

4. What are ext4, NTFS, and FAT32 file systems?

   - A file system is a method used by an Operating System to organize, store, and manage files on a storage device (HDD, SSD, USB drive).

      a. ext4 (Fourth Extended File System)
      Default file system for most Linux operating systems.
      Supports large files and partitions.
      Fast, reliable, and supports journaling (helps recover data after crashes).

      b. NTFS (New Technology File System)
      Default file system for Windows.
      Supports large files, file permissions, encryption, and journaling.
      More secure and efficient than FAT32.
      
      c. FAT32 (File Allocation Table 32)
      An older file system supported by Windows, Linux, and macOS.
      Compatible with most USB drives and memory cards.
      Maximum file size is 4 GB, so it is not suitable for very large files.

5. What is an IP address, subnetting, and CIDR?

      - a. IP Address

      An IP (Internet Protocol) Address is a unique number assigned to every device on a network. It is used to identify devices and enable communication over the internet or a local network.

      Example: 192.168.1.10

      b. Subnetting

      Subnetting is the process of dividing a large network into smaller subnetworks (subnets). It improves network performance, security, and efficient use of IP addresses.

      Example:
      A company divides one network into separate subnets for the HR, Finance, and IT departments.

      c. CIDR (Classless Inter-Domain Routing)

      CIDR is a method of allocating IP addresses using a prefix length instead of fixed classes (A, B, C).

      Example:

      192.168.1.0/24
      /24 means the first 24 bits represent the network, and the remaining 8 bits represent the host.

6. Explain TCP/IP, UDP, ICMP, DNS, DHCP, HTTP/S, and FTP.?

     - a. TCP/IP (Transmission Control Protocol / Internet Protocol)
      TCP ensures reliable and error-free data delivery.
      IP provides the address and route for sending data between devices.
      Used for most internet communication.

        b. UDP (User Datagram Protocol)
        A fast communication protocol that sends data without checking if it is received.
        Faster than TCP but less reliable.
        Used for video streaming, online gaming, and VoIP.

        c. ICMP (Internet Control Message Protocol)
        Used to report network errors and test connectivity.
        Example: The ping command uses ICMP to check if another device is reachable.

        d. DNS (Domain Name System)
        Converts domain names into IP addresses.
        Example: www.google.com → 142.250.xxx.xxx

        e. DHCP (Dynamic Host Configuration Protocol)
        Automatically assigns IP addresses and network settings to devices.
        Eliminates the need to configure IP addresses manually.

        f. HTTP/HTTPS (HyperText Transfer Protocol / Secure)
        HTTP transfers web pages between a browser and a web server.
        HTTPS is the secure version of HTTP that encrypts data using SSL/TLS.
        HTTPS is used for secure websites like online banking and shopping.

        g. FTP (File Transfer Protocol)
        Used to transfer files between a client and a server.
        Commonly used to upload or download website files.
